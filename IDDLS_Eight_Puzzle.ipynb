```python
import copy

def get_entire_grid_input(prompt_name):
    while True:
        user_input = input(f"Enter the 3x3 grid for {prompt_name} as 9 space-separated numbers (e.g., 2 8 3 1 6 4 7 0 5): ").strip()
        parts = user_input.split()
        if len(parts) != 9:
            print("Error: You must enter exactly 9 numbers. Please try again.")
            continue
        try:
            numbers = [int(x) for x in parts]
            # Reconstruct into 3x3 list of lists
            grid = [
                numbers[0:3],
                numbers[3:6],
                numbers[6:9]
            ]
            return grid
        except ValueError:
            print("Error: All entries must be integers. Please try again.")

def dls(board, level, limit, visited):
    global count
    count += 1

    if board == goal:
        print("Goal found")
        print(f"Level: {level}")
        print("Final Board:")
        for row in board:
            print(row)
        return True

    if level >= limit:
        return False

    key = str(board)
    if key in visited and visited[key] <= level:
        return False
    visited[key] = level

    x, y = 0, 0
    for i in range(3):
        for j in range(3):
            if board[i][j] == 0:
                x, y = i, j
                break
        if board[x][y] == 0:
            break

    for dx, dy in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nx, ny = x + dx, y + dy

        if 0 <= nx < 3 and 0 <= ny < 3:
            board[x][y], board[nx][ny] = board[nx][ny], board[x][y]

            if dls(board, level + 1, limit, visited):
                return True

            board[x][y], board[nx][ny] = board[nx][ny], board[x][y]

    return False

def iddfs(start_board):
    global count
    limit = 0
    while True:
        count = 0
        visited = {}
        print(f"\nTrying DLS with limit = {limit}")
        current_board_for_dls = copy.deepcopy(start_board)
        if dls(current_board_for_dls, 0, limit, visited):
            print(f"Total states explored for solution: {count}")
            return True
        print(f"States explored at limit {limit}: {count}")
        limit += 1

start = get_entire_grid_input("Start State")
goal = get_grid_input = get_entire_grid_input("Goal State")

count = 0
print("\nRunning IDDFS...")
iddfs(copy.deepcopy(start))
```
