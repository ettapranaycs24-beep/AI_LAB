```python
import copy

start = [[2, 8, 3],
         [1, 6, 4],
         [7, 0, 5]]

goal = [[1, 2, 3],
        [8, 0, 4],
        [7, 6, 5]]

def dls(board, level, limit, visited):
    global count # Declare count as global
    count += 1 # Increment global count

    if board == goal:
        print("Goal found")
        print(f"Level: {level}")
        print("Final Board:")
        for row in board:
            print(row)
        return True

    if level >= limit:
        return False

    # Use a tuple for visited keys to ensure immutability if board was mutable
    # However, for nested lists, string conversion is a common workaround
    key = str(board)
    if key in visited and visited[key] <= level:
        return False
    visited[key] = level

    x, y = 0, 0
    # Find the blank (0) tile
    for i in range(3):
        for j in range(3):
            if board[i][j] == 0:
                x, y = i, j
                break
        if board[x][y] == 0:
            break

    # Possible moves (up, down, left, right)
    for dx, dy in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nx, ny = x + dx, y + dy

        # Check if the move is within board boundaries
        if 0 <= nx < 3 and 0 <= ny < 3:
            # Make the move
            board[x][y], board[nx][ny] = board[nx][ny], board[x][y]

            # Recursively call dls
            if dls(board, level + 1, limit, visited):
                return True

            # Backtrack (undo the move)
            board[x][y], board[nx][ny] = board[nx][ny], board[x][y]

    return False

def iddfs(start_board):
    global count
    limit = 0
    while True:
        count = 0 # Reset count for each IDDFS iteration
        visited = {}
        print(f"\nTrying DLS with limit = {limit}")
        # Create a deep copy of the board for each DLS call within IDDFS
        current_board_for_dls = copy.deepcopy(start_board)
        if dls(current_board_for_dls, 0, limit, visited):
            print(f"Total states explored for solution: {count}")
            return True
        print(f"States explored at limit {limit}: {count}")
        limit += 1

# Initialize global count (if not already done by kernel)
count = 0

print("Running IDDFS...")
iddfs(copy.deepcopy(start)) # Pass a copy of the start board to IDDFS
```
