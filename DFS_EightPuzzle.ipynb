```python
import copy
from collections import deque

def solve_bfs(start_board, goal_board):
    queue = deque([(start_board, 0)])
    visited = {str(start_board)}

    while queue:
        board, level = queue.popleft()

        if board == goal_board:
            print("\nGoal found!")
            print(f"Level (Moves): {level}")
            print("Final Board State:")
            for row in board:
                print(row)
            return True

        x, y = 0, 0
        for i in range(3):
            for j in range(3):
                if board[i][j] == 0:
                    x, y = i, j
                    break
            else:
                continue
            break

        for dx, dy in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
            nx, ny = x + dx, y + dy

            if 0 <= nx < 3 and 0 <= ny < 3:
                next_board = [row[:] for row in board]
                next_board[x][y], next_board[nx][ny] = next_board[nx][ny], next_board[x][y]

                key = str(next_board)
                if key not in visited:
                    visited.add(key)
                    queue.append((next_board, level + 1))

    print("\nNo solution possible from this start state to the goal state.")
    return False

def get_user_board(prompt_name):
    print(f"Enter the {prompt_name} state (9 numbers from 0 to 8, space-separated, where 0 is the empty tile):")
    while True:
        try:
            user_input = input().strip().split()
            if len(user_input) != 9:
                print("Please enter exactly 9 numbers.")
                continue
            numbers = [int(x) for x in user_input]
            # Structure into a 3x3 grid
            board = [numbers[i:i+3] for i in range(0, 9, 3)]
            return board
        except ValueError:
            print("Invalid input. Please enter numbers only.")

# Take input for start and goal states
start_state = get_user_board("START")
goal_state = get_user_board("GOAL")

print("\nStarting BFS...")
solve_bfs(start_state, goal_state)
```
